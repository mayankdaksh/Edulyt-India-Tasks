## **Here are the solution codes for the given tasks by Edulyt India in Project 1**

### **1. Monthly spend of each customer**

```python
monthly_spend = spend.groupby(['Customer', 'Spend Month'])['Spend Amount'].sum().reset_index()
monthly_spend.columns = ['Customer', 'Spend Month', 'Total Spend']

print("Monthly Spend of Each Customer:")
print(monthly_spend)
```

```python
# Group by 'Customer' to get the total spend for each customer
total_spend_per_customer = spend.groupby('Customer')['Spend Amount'].sum().reset_index()
total_spend_per_customer.columns = ['Customer ID', 'Total Spend']

print("\nTotal Spend of Each Customer:")
print(total_spend_per_customer)
```

```python
# Plotting with Seaborn
plt.figure(figsize=(10, 6))
sns.lineplot(data=monthly_spend, x='Spend Month', y='Total Spend', hue='Customer', marker='o')

plt.title('Monthly Spend per Customer')
plt.xlabel('Month')
plt.ylabel('Total Spend')
plt.legend(title='Customer')
plt.grid(True)
plt.show()
```

### **2. Monthly repayment of each customer**

```python
monthly_repayment = repayment.groupby(['Customer', 'Repayment Month'])['Repayment Amount'].sum().reset_index()
monthly_repayment.columns = ['Customer ID', 'Repayment Month', 'Total Repayment']

print("\nMonthly Repayment of Each Customer:")
print(monthly_repayment)
```

```python
# Group by 'Customer' to get the total repayment for each customer
total_repayment_per_customer = repayment.groupby('Customer')['Repayment Amount'].sum().reset_index()
total_repayment_per_customer.columns = ['Customer ID', 'Total Repayment']

print("\nTotal Repayment of Each Customer:")
print(total_repayment_per_customer)
```

```python
# Plotting with Seaborn
plt.figure(figsize=(10, 6))
sns.lineplot(data=monthly_repayment, x='Repayment Month', y='Total Repayment', hue='Customer ID', marker='o')

plt.title('Monthly Repayment per Customer')
plt.xlabel('Month')
plt.ylabel('Total Repayment')
plt.legend(title='Customer ID')
plt.grid(True)
plt.show()
```

### **3. Highest paying 10 customers**

```python
highest_paying_customers = spend.groupby('Customer')['Spend Amount'].sum().nlargest(10).reset_index()
highest_paying_customers.columns = ['Customer ID', 'Total Spend']
print("\nHighest Paying 10 Customers:")
print(highest_paying_customers)
```

```python
# Plotting using Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(x='Customer ID', y='Total Spend', data=highest_paying_customers, palette='viridis')
plt.xlabel('Customer ID')
plt.ylabel('Total Spend')
plt.title('Highest Paying 10 Customers')
plt.xticks(rotation=45)
plt.show()
```

### **4. People in which segment are spending more money**

```python
spend_with_segment = spend.merge(customer_acqusition[['Customer', 'Segment']], on='Customer')
segment_spending = spend_with_segment.groupby('Segment')['Spend Amount'].sum().reset_index()
segment_spending.columns = ['Segment', 'Total Spend']

print("\nPeople in Which Segment are Spending More Money:")
print(segment_spending)
```

```python
# Plotting using Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(x='Segment', y='Total Spend', data=segment_spending, palette='viridis')
plt.xlabel('Segment')
plt.ylabel('Total Spend')
plt.title('Total Spending Across Customer Segments')
plt.xticks(rotation=45)
plt.show()
```

### **5. Which age group is spending more money?**

```python
# Define age groups
bins = [0, 18, 25, 35, 45, 55, 65, 100]
labels = ['0-18', '19-25', '26-35', '36-45', '46-55', '56-65', '66+']
customer_acqusition['Age'] = pd.cut(customer_acqusition['Age'], bins=bins, labels=labels, right=False)
spend_with_age_group = spend.merge(customer_acqusition[['Customer', 'Age']], on='Customer')
age_group_spending = spend_with_age_group.groupby('Age')['Spend Amount'].sum().reset_index()
age_group_spending.columns = ['Age Group', 'Total Spend']

print("\nWhich Age Group is Spending More Money:")
print(age_group_spending)
```

```python
# Plotting using Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(x='Age Group', y='Total Spend', data=age_group_spending, palette='viridis')
plt.xlabel('Age Group')
plt.ylabel('Total Spend')
plt.title('Total Spending Across Age Groups')
plt.xticks(rotation=45)
plt.show()
```

### **6. Which is the most profitable segment?**

```python
# Profit is spend - repayment
profit_data = spend.merge(repayment, on='Customer', suffixes=('_spend', '_repayment'))
profit_data['profit'] = profit_data['Spend Amount'] - profit_data['Repayment Amount']
profit_with_segment = profit_data.merge(customer_acqusition[['Customer', 'Segment']], on='Customer')
segment_profit = profit_with_segment.groupby('Segment')['profit'].sum().reset_index()
segment_profit.columns = ['Segment', 'Total Profit']

print("\nMost Profitable Segment:")
print(segment_profit)
```

```python
# Plotting with Seaborn
plt.figure(figsize=(8, 6))
sns.barplot(x='Segment', y='Total Profit', data=segment_profit, palette='viridis')
plt.xlabel('Segment')
plt.ylabel('Total Profit')
plt.title('Total Profit Across Customer Segments')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### **7. In which category the customers are spending more money?**

```python
category_spending = spend.groupby('Type')['Spend Amount'].sum().reset_index()
category_spending.columns = ['Category', 'Total Spend']

print("\nCategory in Which Customers are Spending More Money:")
print(category_spending)
```

```python
# Plotting with Seaborn
plt.figure(figsize=(10, 6))
sns.barplot(x='Category', y='Total Spend', data=category_spending, palette='viridis')
plt.xlabel('Category')
plt.ylabel('Total Spend')
plt.title('Total Spend Across Categories (Seaborn)')
plt.xticks(rotation=45)
plt.show()
```

### **8. Impose an interest rate of 2.9% for each customer for any due amount.**

```python
# Group by customer and month, then sum the amounts
monthly_repayment = repayment.groupby(['Customer', 'Repayment Month'])['Repayment Amount'].sum().reset_index()
monthly_spend = spend.groupby(['Customer', 'Spend Month'])['Spend Amount'].sum().reset_index()

# Merge the spend and repayment data on 'Customer' and 'Month'
monthly_financials = pd.merge(monthly_spend, monthly_repayment, how='left', left_on=['Customer', 'Spend Month'], right_on=['Customer', 'Repayment Month']).fillna(0)

# Calculate due amount and interest
monthly_financials['Due Amount'] = monthly_financials['Spend Amount'] - monthly_financials['Repayment Amount']
monthly_financials['Interest'] = monthly_financials['Due Amount'].apply(lambda x: x * 0.029 if x > 0 else 0)

# Drop the 'Repayment Month' column as it's redundant now
monthly_financials.drop(columns=['Repayment Month'], inplace=True)

print(monthly_financials)
```

### **9. Monthly profit for the bank**

```python
monthly_financials['Profit'] = monthly_financials['Due Amount'] + monthly_financials['Interest']
monthly_profit = monthly_financials.groupby('Spend Month')['Profit'].sum().reset_index()
monthly_profit.columns = ['Month', 'Total Profit']

print("\nMonthly Profit for the Bank:")
print(monthly_profit)
```

```python
# Plotting with Matplotlib
plt.figure(figsize=(10, 6))
plt.plot(monthly_profit['Month'], monthly_profit['Total Profit'], marker='o', linestyle='-', color='blue')

plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.title('Monthly Profit for the Bank')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
